```python
from pathlib import Path
from ultralytics import YOLO

DATA = '/kaggle/input/soccernet5-yolo/data.yaml'
WEIGHTS = '/kaggle/input/soccernet5-yolo/pretrained_best.pt'
print('data', Path(DATA).exists(), 'weights', Path(WEIGHTS).exists())
assert Path(DATA).exists() and Path(WEIGHTS).exists()

model = YOLO(WEIGHTS)
results = model.train(
    data=DATA,
    epochs=50,
    imgsz=960,
    batch=8,
    device=0,
    project='/kaggle/working/runs',
    name='kaggle_finetune_v9c',
    exist_ok=True,
    patience=12,
    workers=2,
    optimizer='AdamW',
    lr0=0.001,
    lrf=0.01,
    cos_lr=True,
    close_mosaic=10,
    save=True,
    save_period=5,
    plots=True,
)
print('DONE', getattr(results, 'save_dir', '/kaggle/working/runs'))
```
